Task 1 (Understand): Load any PDF with both mode='page' and mode='single'. Print len(documents) and documents[0].metadata for both. Observe the difference.

```python
from pathlib import Path
from langchain_community.document_loaders import PyPDFLoader
from langchain_community.document_loaders import PDFMinerLoader
```

```python
file_path = Path("../files/Mastering RAG-compressed.pdf")
file_path.exists()  # Check if the file exists
```

#### mode = page ---> it will load the document - page wise ---- number of page in pdf == number of documents

```python
loader = PyPDFLoader(file_path = file_path.as_posix(),mode="page")
```

```python
docs = loader.load()
```

```python
len(docs)  # Check the number of documents loaded
```

```python
print(docs[55].page_content)  # Print the content of the first document
```

```python
pdfminer_loader = PDFMinerLoader(file_path = file_path.as_posix(),mode="page")
```

```python
pdfminer_docs = pdfminer_loader.load()
```

```python
len(pdfminer_docs)
```

#### Chaning mode to single  --- > It will load the pdf as a single document .

```python
singlepdfloader = PyPDFLoader(file_path = file_path.as_posix(),mode="single")
```

```python
singlepdf_docs = singlepdfloader.load()
```

```python
len(singlepdf_docs)
```
